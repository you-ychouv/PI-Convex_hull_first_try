# In this notebook, we will compare the performance of these two algorithms on different datasets

```python
import numpy as np
import matplotlib.pyplot as plt
import Sweeping
import Output_Sensitive
import Dataset
import time
```

```python
convexHullNaive = Sweeping.sweeping
convexHull = Output_Sensitive.convexHull
datasetA = Dataset.datasetA
datasetB = Dataset.datasetB
datasetC = Dataset.datasetC
datasetD = Dataset.datasetD
```

### Dataset A

```python
point = datasetA(1000)
t0 = time.time()
convex1 = np.array(convexHullNaive(point))
t1 = time.time()
print(f"Total runtime of Sweeping algorithm is {t1 - t0} seconds")



t0 = time.time()
convex1 = np.array(convexHull(point))
t1 = time.time()
print(f"Total runtime of Output sensitive algorithm is {t1 - t0} seconds")
```

```python
def plotting(point):
    t0 = time.perf_counter()
    convex1 = np.array(convexHullNaive(point))
    t1 = time.perf_counter()
    print(f"Total runtime of Sweeping algorithm is {t1 - t0} seconds")

    t2 = time.perf_counter()
    convex2 = convexHull(point)
    t3 = time.perf_counter()
    print(f"Total runtime of the Output Sensitive algorithm is {t3 - t2} seconds")

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
    ax1.scatter([p[0] for p in point], [p[1] for p in point], alpha=0.3)
    ax1.plot(convex1[:, 0], convex1[:, 1], 'r-o')
    ax1.set_title('Sweeping Algorithm')
    ax1.set_xlabel('x-axis')
    ax1.set_ylabel('y-axis')
    ax1.set_aspect('equal', adjustable='box')

    ax2.scatter([p[0] for p in point], [p[1] for p in point], alpha=0.3)
    ax2.plot([p[0] for p in convex2], [p[1] for p in convex2], 'r-o')
    ax2.set_title('Output Sensitive Algorithm')
    ax2.set_xlabel('x-axis')
    ax2.set_ylabel('y-axis')
    ax2.set_aspect('equal', adjustable='box')
    plt.suptitle("Comparison")
    plt.tight_layout()
    plt.show()
```

```python
point1 = datasetB(100000)
t0 = time.perf_counter()
convex1 = np.array(convexHullNaive(point1))
t1 = time.perf_counter()
print(f"Total runtime of Sweeping algorithm is {t1 - t0} seconds")
```

```python
t2 = time.perf_counter()
convex2 = convexHull(point1)
t3 = time.perf_counter()
print(f"Total runtime of the Output Sensitive algorithm is {t3 - t2} seconds")
```

```python
plotting(datasetA(1000))
```
